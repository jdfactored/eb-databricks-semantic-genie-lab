# 99 — Clean Up

**Run this only after your work has been graded.**

It permanently drops your lab schema with every table and view in it (synthetic data, the metric view,
your observations), and moves the Genie Agents you registered to the trash. There is no undo for the schema.

Not touched, remove by hand if you created them: draft Pages (**Discover > Pages**), Genie One chats,
scheduled tasks and skills, and any unregistered agent clones.

```python
from helpers import utils, setup_schemas, genie_agents

cfg = utils.get_configs()
print("This will permanently delete:")
print(f"  {cfg['catalog']}.{cfg['schema']}  (schema, CASCADE)")
for a in genie_agents.registered_agents():
    print(f"  Genie Agent '{a['title']}' ({a['role']}, {a['space_id']})  -> trash")
```

## Confirm

```python
CONFIRM = False #Change it to True to remove all the assets used
TRASH_GENIE_AGENTS = True

if CONFIRM:
    if TRASH_GENIE_AGENTS:
        for a in genie_agents.registered_agents():
            try:
                genie_agents.trash(a["space_id"])
                print(f"Trashed Genie Agent '{a['title']}'")
            except Exception as exc:
                print(f"Could not trash '{a['title']}' ({type(exc).__name__}); delete it in the UI.")
    setup_schemas.clean_up_schema()
    print("\nCleanup complete.")
else:
    print("Nothing deleted. Set CONFIRM = True to proceed.")
```
